# Mini-Keras MNIST End-to-End Demo

This notebook demonstrates building, training, saving, loading, and evaluating a neural network using **Mini-Keras** (built from scratch with NumPy).

## 1. Imports

```python
import numpy as np
import matplotlib.pyplot as plt
from mini_keras import Sequential, Dense
from mini_keras.activations import ReLU, Softmax
from mini_keras.datasets import load_mnist, visualize_mnist_images
from mini_keras.utils import to_categorical
```

## 2. Load and Preprocess Data

```python
# Load MNIST dataset (downloaded automatically if needed)
(X_train, y_train), (X_test, y_test) = load_mnist()

# Flatten images from (N, 28, 28) to (N, 784) and normalize pixels to [0, 1]
X_train_flat = X_train.reshape(-1, 28 * 28).astype('float32') / 255.0
X_test_flat = X_test.reshape(-1, 28 * 28).astype('float32') / 255.0

# One-hot encode targets
Y_train = to_categorical(y_train, num_classes=10)
Y_test = to_categorical(y_test, num_classes=10)

print(f"X_train shape: {X_train_flat.shape}, Y_train shape: {Y_train.shape}")
print(f"X_test shape:  {X_test_flat.shape},  Y_test shape:  {Y_test.shape}")
```

## 3. Define and Train Model

```python
model = Sequential()
model.add(Dense(units=128, input_shape=(784,)))
model.add(ReLU())
model.add(Dense(units=64))
model.add(ReLU())
model.add(Dense(units=10))
model.add(Softmax())

model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

# Train for 5 epochs
history = model.fit(X_train_flat, Y_train, epochs=5, batch_size=64)
```

## 4. Save and Load Model

```python
# Save model to disk
model.save_model('model/mnist.ab')
print('Model saved to model/mnist.ab')

# Load model back
loaded_model = Sequential.load_model('model/mnist.ab')
print('Model successfully loaded back!')
```

## 5. Inference and Visualization (Replacing trysavemodel.py)

```python
# Make predictions on a test sample
sample_index = 4
sample_data = np.reshape(X_test_flat[sample_index], (1, -1))
prediction = loaded_model.predict(sample_data)
predicted_label = np.argmax(prediction)
true_label = y_test[sample_index]

print(f"Predicted label: {predicted_label}")
print(f"True label:      {true_label}")

# Display test image
plt.figure(figsize=(4, 4))
plt.imshow(X_test[sample_index], cmap='gray')
plt.title(f"True Label: {true_label} - Predicted: {predicted_label}")
plt.axis('off')
plt.show()
```

## 6. Visualize Sample Predictions

```python
digit_labels = [str(i) for i in range(10)]
visualize_mnist_images(X_train_flat, Y_train, digit_labels, num_images=10)
```
